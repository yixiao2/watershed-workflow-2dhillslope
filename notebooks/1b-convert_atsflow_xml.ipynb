# convert ATS flow xml from ATS 1.5 to ATS 1.6
- Input: xml files in folders `caseflow-run0`, `caseflow-run1`, and `caseflow-run2`
- Output: xml files in the same folders with suffix `*v1.6.xml`
- Notes
    - ATS-PFLOTRAN xml files are generated based on flow xml files for ATS 1.6

## Parameters and data sources

```python
from config_utils import load_config

config = load_config('config.json')
site_name = config['case']['site_name']
```

```python
from scipy.io import loadmat
m2_mat_filename =  f'../data-processed/{site_name}/m2_coords_{site_name}.mat'
loaded_data = loadmat(m2_mat_filename)
meshsize_nx = loaded_data['meshsize_nx'].flatten()[0]
dzs_soil  = loaded_data['dzs_soil'].flatten()
dzs_geo   = loaded_data['dzs_geo'].flatten()
#m2_coords = loaded_data['m2_coords']
```

## for ATS-flow run0/1/2

```python
import os
import sys
from pathlib import Path
import subprocess
```

```python
## NERSC
old_amanzi_path = '/global/cfs/cdirs/m1800/xiaoyi/softwares/ats-152-250212/repos/amanzi/tools/amanzi_xml'
new_amanzi_path = '/global/cfs/cdirs/m1800/xiaoyi/softwares/ats-master-260109/repos/amanzi/tools/amanzi_xml'
xmlinputconvert_path = Path('/global/cfs/cdirs/m1800/xiaoyi/softwares/ats-master-260109/repos/amanzi/src/physics/ats/tools/input_converters')
```

```python
# Remove old amanzi_xml path if it exists
if old_amanzi_path in sys.path:
    sys.path.remove(old_amanzi_path)

# Add new amanzi_xml path at the beginning (highest priority)
sys.path.insert(0, str(new_amanzi_path))

# Verify the change
import amanzi_xml
print(f"Using amanzi_xml from: {amanzi_xml.__file__}")
```

```python
base_dir = Path().resolve().parent  # Assumes the notebook is inside the notebooks/ directory

# Create a clean environment and set only the new PYTHONPATH
env = os.environ.copy()

# REPLACE (not append) the PYTHONPATH to ensure old path isn't used
env['PYTHONPATH'] = new_amanzi_path
```

```python
## caseflow-run0
tmp_xml = base_dir / 'caseflow-run0' / f'{site_name}_nx{meshsize_nx}_nz{len(dzs_soil)+len(dzs_geo)}.run0.v1.5.xml'
tmp_xml_o = base_dir / 'caseflow-run0' / f'{site_name}_nx{meshsize_nx}_nz{len(dzs_soil)+len(dzs_geo)}.run0.v1.6.xml'
command = f"python {xmlinputconvert_path / 'xml-1.5-1.6.py'} {tmp_xml} -o {tmp_xml_o}"
try:
    result = subprocess.run(command, shell=True, check=True, env=env, 
                          capture_output=True, text=True)
    print(result.stdout)
except subprocess.CalledProcessError as e:
    print(f"Error occurred: {e}")

## caseflow-run1
tmp_xml = base_dir / 'caseflow-run1' / f'{site_name}_nx{meshsize_nx}_nz{len(dzs_soil)+len(dzs_geo)}.run1.v1.5.xml'
tmp_xml_o = base_dir / 'caseflow-run1' / f'{site_name}_nx{meshsize_nx}_nz{len(dzs_soil)+len(dzs_geo)}.run1.v1.6.xml'
command = f"python {xmlinputconvert_path / 'xml-1.5-1.6.py'} {tmp_xml} -o {tmp_xml_o}"
try:
    result = subprocess.run(command, shell=True, check=True, env=env, 
                          capture_output=True, text=True)
    print(result.stdout)
except subprocess.CalledProcessError as e:
    print(f"Error occurred: {e}")

## caseflow-run2
tmp_xml = base_dir / 'caseflow-run2' / f'{site_name}_nx{meshsize_nx}_nz{len(dzs_soil)+len(dzs_geo)}.run2.v1.5.xml'
tmp_xml_o = base_dir / 'caseflow-run2' / f'{site_name}_nx{meshsize_nx}_nz{len(dzs_soil)+len(dzs_geo)}.run2.v1.6.xml'
command = f"python {xmlinputconvert_path / 'xml-1.5-1.6.py'} {tmp_xml} -o {tmp_xml_o}"
try:
    result = subprocess.run(command, shell=True, check=True, env=env, 
                          capture_output=True, text=True)
    print(result.stdout)
except subprocess.CalledProcessError as e:
    print(f"Error occurred: {e}")
```
