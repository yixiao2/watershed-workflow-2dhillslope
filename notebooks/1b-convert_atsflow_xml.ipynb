- Output: xml files in the same folders with suffix `*v1.6.xml`

```python
import os
import sys
from pathlib import Path
import subprocess

# Remove old amanzi_xml path if it exists
old_amanzi_path = '/home/xiao284/softwares/ats-15-25jan/repos/amanzi/tools/amanzi_xml'
if old_amanzi_path in sys.path:
    sys.path.remove(old_amanzi_path)

# Add new amanzi_xml path at the beginning (highest priority)
new_amanzi_path = '/home/xiao284/softwares/ats-master-251124/repos/amanzi/tools/amanzi_xml'
## phong's branch uses Path('/home/xiao284/softwares/ats-15-25sep-phong/repos/amanzi/src/physics/ats/tools/input_converters/xml-1.5-master.py')
sys.path.insert(0, str(new_amanzi_path))

# Verify the change
import amanzi_xml
print(f"Using amanzi_xml from: {amanzi_xml.__file__}")
```

```python
base_dir = Path().resolve().parent  # Assumes the notebook is inside the notebooks/ directory

xmlinputconvert_path = Path('/home/xiao284/softwares/ats-master-251124/repos/amanzi/src/physics/ats/tools/input_converters')

# Create a clean environment and set only the new PYTHONPATH
env = os.environ.copy()
new_pythonpath = '/home/xiao284/softwares/ats-master-251124/repos/amanzi/tools/amanzi_xml'
# REPLACE (not append) the PYTHONPATH to ensure old path isn't used
env['PYTHONPATH'] = new_pythonpath
```

```python
## caseflow-run0
tmp_xml = base_dir / 'caseflow-run0' / 'NF01_nx100_nz18.run0.v1.5.xml'
tmp_xml_o = base_dir / 'caseflow-run0' / 'NF01_nx100_nz18.run0.v1.6.xml'
command = f"python {xmlinputconvert_path / 'xml-1.5-1.6.py'} {tmp_xml} -o {tmp_xml_o}"
try:
    result = subprocess.run(command, shell=True, check=True, env=env, 
                          capture_output=True, text=True)
    print(result.stdout)
except subprocess.CalledProcessError as e:
    print(f"Error occurred: {e}")

## caseflow-run1
tmp_xml = base_dir / 'caseflow-run1' / 'NF01_nx100_nz18.run1.v1.5.xml'
tmp_xml_o = base_dir / 'caseflow-run1' / 'NF01_nx100_nz18.run1.v1.6.xml'
command = f"python {xmlinputconvert_path / 'xml-1.5-1.6.py'} {tmp_xml} -o {tmp_xml_o}"
try:
    result = subprocess.run(command, shell=True, check=True, env=env, 
                          capture_output=True, text=True)
    print(result.stdout)
except subprocess.CalledProcessError as e:
    print(f"Error occurred: {e}")

## caseflow-run2
tmp_xml = base_dir / 'caseflow-run2' / 'NF01_nx100_nz18.run2.v1.5.xml'
tmp_xml_o = base_dir / 'caseflow-run2' / 'NF01_nx100_nz18.run2.v1.6.xml'
command = f"python {xmlinputconvert_path / 'xml-1.5-1.6.py'} {tmp_xml} -o {tmp_xml_o}"
try:
    result = subprocess.run(command, shell=True, check=True, env=env, 
                          capture_output=True, text=True)
    print(result.stdout)
except subprocess.CalledProcessError as e:
    print(f"Error occurred: {e}")
```
